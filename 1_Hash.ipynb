```python
import hashlib

# 1. Define the input string
text = "hello"

# 2. Convert text to bytes and hash it
hash_object = hashlib.sha256(text.encode())

print(text.encode())

# 3. Get the 64-character hexadecimal string
hex_dig = hash_object.hexdigest()

print("Hex_dig:",hex_dig,"its length is ",len(hex_dig))
# Output: 2cf24dbc5c0b2dd1b06865b8bb99a011340b410752706b989bc11e247b493430
```
